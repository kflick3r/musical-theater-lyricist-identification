## Related Work

For my project, I selected five research papers that explore different approaches to authorship attribution, stylometry, and the analysis of song lyrics.

1. *Song Authorship Attribution: A Lyrics and Rhyme Based Approach* (2023)
2. *Not All Character N-grams Are Created Equal* (2015)
3. *Attributing Authorship via the Perplexity of Authorial Language Models* (2025)
4. *Genre-Conformity in the Topics of Lyrics and Song Popularity* (2024)
5. *Topic or Style? Exploring the Most Useful Features for Authorship Attribution* (2018)


### 1. *Song Authorship Attribution: A Lyrics and Rhyme Based Approach* (2023)

Yılmaz and Scheffler explored whether song authors could be identified through patterns in their lyrics, including word structure, pronunciation, and rhyme. They created a balanced dataset of 12,000 songs from 120 artists, with 100 songs per artist. Using Convolutional Neural Networks (CNNs), they compared three types of representations: character embeddings, subword embeddings, and phoneme representations designed to capture pronunciation patterns such as rhyme and repetition. Of the individual representations, character embeddings performed best, achieving 23.6% accuracy using 10-fold cross-validation. A model combining all three representations achieved 29.8% accuracy on a fixed train/test split. The researchers also used occlusion analysis to investigate which parts of the lyrics influenced predictions and found that models sometimes relied on distinctive words rather than broader stylistic patterns.

One important difference between this research and my project is that Yılmaz and Scheffler treated performers as the authors of their songs, even though performers do not always write their own lyrics. I will need to be careful about identifying the actual writers rather than the performers. I also plan to exclude songs with ambiguous or shared authorship whenever possible. Their dataset preparation gave me a few things to consider, particularly removing duplicate lyrics and keeping the number of songs per lyricist reasonably balanced. I also found their use of occlusion analysis interesting because it showed how much individual words could influence a model's predictions. Even though I won't be using their CNN architecture, I think this paper provides a useful starting point for building my dataset and understanding some of the challenges of classifying song lyrics.

### 2. *Not All Character N-grams Are Created Equal* (2015)

Character n-grams are commonly used for authorship attribution, but Sapkota et al. wanted to understand whether certain types were more useful than others. They divided character n-grams into ten categories based on the linguistic information they capture, including prefixes, suffixes, punctuation patterns, and parts of words. The researchers evaluated these features in both single-domain and cross-domain settings to see how well they performed when authors wrote about similar or different topics. They found that affix and punctuation features were particularly effective, while n-grams associated with specific words were often less useful. In some cases, removing word-related n-grams maintained or even improved classification performance while reducing the number of features needed.

I originally planned to compare word-level TF-IDF with Logistic Regression against pretrained Transformer embeddings with Logistic Regression, but this paper gives me a reason to also experiment with character-level TF-IDF. I hadn't really considered how much information could be captured by small sequences of characters, especially patterns involving prefixes, suffixes, and punctuation. I'm curious whether these features will be useful for distinguishing lyricists based on their writing habits rather than just the words they use. The comparison between single-domain and cross-domain classification was also interesting because I plan to test whether my models can identify lyricists across different musicals. This gives me another way to think about which features might generalize beyond the songs used for training.

### 3. *Attributing Authorship via the Perplexity of Authorial Language Models* (2025)

Rather than using a traditional classification model, Huang et al. approached authorship attribution by measuring how well language models could predict an author's writing. They created a separate Authorial Language Model (ALM) for each candidate author by further training a pretrained language model on samples of that author's work. When identifying an unknown text, they calculated its perplexity using each ALM and assigned it to the author whose model produced the lowest score. The researchers tested this approach on four datasets containing blogs, news articles, newspaper columns, and movie reviews. Their method achieved an average macro-accuracy of 88.1% across the four datasets, including 99.5% accuracy on the movie review dataset. Another interesting finding was that content words often provided more information about authorship than function words, challenging traditional assumptions about which linguistic features are most useful for identifying authors.

I found this paper interesting because it approaches authorship attribution differently from the classification methods I'm planning to use. I hadn't considered using perplexity for authorship attribution before reading this paper, so I was surprised by how well it performed across different datasets. I was also surprised by their finding that content words could be more informative than function words, since writing style is often associated with things like grammar and word usage rather than subject matter. I don't plan to implement their approach because it would require training separate language models for my lyricists, but it helped me better understand how pretrained language models can be adapted to identify individual authors.

### 4. *Genre-Conformity in the Topics of Lyrics and Song Popularity* (2024)

Aljanaki investigated how musical genres influence the topics discussed in song lyrics and whether songs that follow typical genre conventions are more popular. Using the Music4All dataset of over 109,000 songs, the researcher examined the relationship between lyrical themes, musical genres, and listening popularity. This study grouped genre tags into 15 categories and used pretrained sentence embeddings to organize lyrics into 36 topics. Some genres, particularly hip-hop and death metal, were strongly associated with specific lyrical themes, while genres such as rock and pop covered a much wider range of topics. The results also showed that songs with lyrics that differed from their genre's typical themes were generally more popular than songs that followed genre conventions, although the relationship varied across genres.

This paper is a little different from the others because it focuses on genre and popularity rather than authorship attribution. However, I think it raises an important question about how much the content of a song is influenced by its genre. This is especially interesting for musical theater, where lyrics are written for specific characters, situations, and storylines. A lyricist might write a comedic song for one character and an emotional ballad for another, even within the same musical. Because of this, songs written by the same person might have very different vocabulary and themes. This paper made me think more carefully about how differences in characters and storylines might affect the vocabulary my models learn, even when the songs are written by the same lyricist.

### 5. *Topic or Style? Exploring the Most Useful Features for Authorship Attribution* (2018)

One challenge in authorship attribution is determining whether a model is recognizing an author's writing style or simply identifying the topics they tend to write about. Sari et al. examined this problem by comparing style-based, content-based, and hybrid features across four datasets containing news articles, legal judgments, and movie reviews. Using topic modeling and feature ablation experiments, they analyzed which features contributed most to classification accuracy under different conditions. Their results showed that content-based features tended to work better when authors wrote about different topics, while style-based features were more useful when authors discussed similar subjects. They also found that character n-grams were not equally effective across all datasets. By incorporating these findings into a neural authorship attribution model, the researchers improved upon previous results on two of the four datasets. On the IMDb62 movie review dataset, their model achieved 95.93% accuracy using additional style-based features, compared to 94.80% for the baseline model.

This paper connects directly to one of my biggest concerns about my project: whether my models will actually learn each lyricist's writing style or simply recognize vocabulary associated with particular musicals. I found it interesting that the researchers couldn't identify one type of feature that worked best across every dataset. Instead, the usefulness of different features depended on how much the authors' topics varied. This is something I want to investigate by comparing random song-level train/test splits against an evaluation where entire musicals are excluded from training. If a model performs well on random splits but struggles with songs from unfamiliar musicals, that could suggest it was relying on vocabulary or other patterns associated with specific productions. Even if a model performs well on both evaluations, I'll still need to consider whether historical differences between my lyricists are influencing its predictions.
